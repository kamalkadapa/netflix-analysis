```python
from datetime import datetime

current_year = datetime.now().year

df["content_age"] = current_year - df["release_year"]
```

```python
df["decade"] = (
    (df["release_year"] // 10) * 10
).astype(str) + "s"
```

```python
df = df.dropna(subset=["duration"])
```

```python
import pandas as pd

df["duration_value"] = pd.to_numeric(
    df["duration"].str.extract(r'(\d+)')[0],
    errors="coerce"
)

def duration_type(row):
    if pd.isna(row["duration_value"]):
        return "Unknown"

    if row["type"] == "Movie":
        if row["duration_value"] <= 60:
            return "Short"
        elif row["duration_value"] <= 120:
            return "Medium"
        else:
            return "Long"
    else:
        if row["duration_value"] == 1:
            return "Short"
        elif row["duration_value"] <= 3:
            return "Medium"
        else:
            return "Long"

df["duration_type"] = df.apply(duration_type, axis=1)
```

```python
df["Weekend Release"] = df["date_added"].dt.dayofweek.apply(
    lambda x: "Yes" if x in [5, 6] else "No"
)
```

```python
df["cast_count"] = df["cast"].fillna("").apply(
    lambda x: len(x.split(",")) if x != "" else 0
)
```

```python
df["Genre Count"] = df["listed_in"].fillna("").apply(
    lambda x: len(x.split(",")) if x != "" else 0
)
```
